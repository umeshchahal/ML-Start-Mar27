```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
normal = np.random.normal(0,1,1000)
right_skew = np.random.exponential(1,1000)
left_skew = -np.random.exponential(1,1000)
uniform = np.random.uniform(0,1,1000)
```

```python
sns.histplot(normal, kde=True)
plt.title("Normal Distribution")
plt.show()
```

```python
sns.histplot(right_skew,kde=True)
plt.title("Right Skewed")
plt.show()
```

```python
sns.histplot(left_skew, kde=True)
plt.title("left_skew")
plt.show()
```

```python
sns.histplot(uniform, kde= True)
plt.title("Uniform")
plt.show()
```
